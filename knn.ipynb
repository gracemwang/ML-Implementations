```python
import numpy as np
from dataclasses import dataclass

@dataclass
class KNN:
    k: int
    features: np.ndarray
    labels: np.ndarray

    def predict(self, features):
        num_samples, _ = features.shape
        predictions = np.empty(num_samples)

        for idx, feature in enumerate(features):
            dist = np.linalg.norm
            distances = [dist(feature - train_feat) for train_feat in self.features]
            dist_idxs = np.argsort(distances)[:self.k]

            common_l = np.bincount([self.labels[idx] for idx in dist_idxs]).argmax()

            predictions[idx] = common_l


        return predictions
```

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split

iris = datasets.load_iris()

x_train, x_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size=0.25)

knn = KNN(k=2, features=x_train, labels=y_train)

preds = knn.predict(x_test)

accuracy = np.mean(preds == y_test)

print(accuracy)
```
