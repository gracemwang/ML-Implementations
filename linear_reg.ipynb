```python
import numpy as np
from dataclasses import dataclass

"""
y = xB + b

MSELoss = sum (actual - predicted)^2 / num_samples

wrt weights
2 * (y - yHat) * (-x) / num_samples

wrt bias
2 * sum (y - yHat) * (-1) / num_samples
"""


@dataclass
class LinearRegression:
    features: np.ndarray
    labels: np.ndarray
    learning_rate: float
    epochs: int
    logging: bool

    def fit(self) -> None:
        """Fits the Linear Regression model"""

        num_samples, num_features = self.features.shape

        self.weights, self.bias = np.zeros(num_features), 0

        for _ in range(self.epochs):
            residuals = self.labels - self.predict(self.features)

            d_weights = (residuals @ (-2*self.features)) / num_samples
            d_bias = 2*residuals.sum()/num_samples

            self.weights -= self.learning_rate * d_weights
            self.bias -= self.learning_rate * d_bias

            if self.logging:
                print(f"MSE Loss [{self.epoch}]: {np.mean(np.square(residuals)):.3f}")



    def predict(self, features:np.ndarray) -> np.ndarray:
        """Performs inference"""

        return features @ self.weights + self.bias
```
